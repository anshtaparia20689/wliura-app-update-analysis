```r
df <- read.csv("app_verisons.csv", stringsAsFactors = FALSE, check.names = FALSE)
```

```r
names(df) <- trimws(names(df))
```

```r
category_counts <- as.data.frame(table(df$`Standardized Update Category`))

names(category_counts) <- c("Standardized Update Category", "Count")

category_counts <- category_counts[order(-category_counts$Count), ]
rownames(category_counts) <- NULL

category_counts
```

```r
df$Year <- format(as.Date(df$`Version Release / Update Date`), "%Y")

category_by_year <- as.data.frame(
  table(df$Year, df$`Standardized Update Category`)
)

names(category_by_year) <- c("Year", "Standardized Update Category", "Count")

category_by_year <- category_by_year[category_by_year$Count > 0, ]

category_by_year <- category_by_year[order(category_by_year$Year, -category_by_year$Count), ]

rownames(category_by_year) <- NULL

category_by_year
```

```r
most_common_by_year <- category_by_year[
  ave(category_by_year$Count, category_by_year$Year, FUN = max) == category_by_year$Count,
]

rownames(most_common_by_year) <- NULL

most_common_by_year
```
